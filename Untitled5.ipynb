```python
from clm import CLMClient, Choice, Noul, Score
import time
start=time.time()
client = CLMClient(
    base_url="http://0.0.0.0:8700"
)                          # CLM_BASE_URL (default http://127.0.0.1:8700), CLM_API_KEY
r = client.system_one(
    state = """
Agent: Thank you for calling PayFlow merchant support, this is Ravi. How can I help?

Customer: Finally, a human. This is my third call in two days. My settlement for the 24th
was supposed to hit my bank account yesterday. It's still showing "On Hold" in the dashboard.

Agent: I'm sorry about that. Can I have your merchant ID?

Customer: M-48213. And before you ask, yes, my KYC is complete. I re-uploaded the GST
certificate last week because your system said it was "expired". It wasn't expired.

Agent: I see the hold. It looks like the re-verification is still pending review.

Customer: Pending for a WEEK? I have 40 vendors waiting on payouts. And on top of that,
my webhooks have been failing since Monday, so my order system thinks payments never went
through. Customers are getting double-charged because my retry logic is firing. Now I'm
getting chargeback notifications for orders I already fulfilled!

Agent: I understand. Let me check whether the webhook failures are linked to the hold...

Customer: Don't "check", fix it. And when I tried to log into the risk console to dispute
these chargebacks, the 2FA code never arrives. I'm locked out of the one place I can
actually respond to them. The dispute window closes in 48 hours. If I lose these, I'm
filing a complaint with the RBI ombudsman and moving to another.
""",
questions = {
    "urgency": Noul(
        instructions="Is this urgent? Consider deadlines, financial exposure, and escalation threats."
    ),
    "is_merchant": Noul(instructions="Is the the customer a merchant or not?"),
    "affects_settlements": Noul(instructions="Does the customer report delayed or held payouts?"),
  "affects_compliance":  Noul(instructions="Does the issue involve KYC or document verification?"),
  "affects_disputes":    Noul(instructions="Does the customer have chargebacks or dispute deadlines at risk?"),
  "affects_integration": Noul(instructions="Does the customer report webhook, API, or retry failures?"),
  "affects_access":      Noul(instructions="Does the customer report login or 2FA/OTP problems?"),
    "subdomain": Choice(
        instructions="Pick the most specific root-cause subdomain driving the customer's main complaint.",
        criteria={
            "kyc_reverification_backlog": "Payout hold caused by pending document re-review",
            "false_document_expiry_flag": "Valid document incorrectly marked as expired",
            "webhook_delivery_failure": "Callbacks failing or timing out",
            "duplicate_charge_from_retries": "Merchant retries causing double debits",
            "chargeback_deadline_risk": "Dispute window closing with no console access",
            "otp_delivery_failure": "2FA codes not delivered",
        },
    ),
    "secondary_domains": Choice(
        instructions="Which other domains are also affected? Select all that apply.",
        criteria={
            "settlements_payouts": "Payouts delayed",
            "compliance_kyc": "Verification issues",
            "disputes_chargebacks": "Chargeback exposure",
            "integration_api": "Webhook/retry problems",
            "account_access": "Login/2FA issues",
        },
    ),
    "needs_risk_ops":   Noul(instructions="Does this need Risk Ops (KYC holds, chargebacks, disputes)?"),
"needs_platform_eng":   Noul(instructions="Does this need Platform Engineering (webhooks, OTP delivery, API failures)?"),
"needs_account_mgmt":   Noul(instructions="Is this a high-value merchant at churn risk needing an account manager?"),
    "frustration": Score(
        instructions="How frustrated is the customer? Consider repeat contacts, tone, and threats.",
        criteria=["Extremely pleased", "Calm", "Frustrated", "Very angry"],
    ),
    "churn_risk": Noul(
        instructions="Is the customer threatening to leave or escalate to a regulator?"
    ),
    "escalation_target": Choice(
        instructions="Which team should own this ticket first?",
        criteria={
            "risk_ops": "Compliance holds, KYC, chargebacks",
            "platform_engineering": "Webhooks, OTP delivery, API failures",
            "account_management": "High-value merchants at churn risk",
        },
    ),
},
)
end=time.time()
print(end-start)
print(r.answers["urgency"].noul)                # 0.41022     probability the statement is true
print(r.answers)     # 38 58.1     (106 tokens on a cold cache: option texts are embedded once)
```

```python
from clm import CLMClient, Choice, Noul, Score
import time
start=time.time()
client = CLMClient(
    base_url="http://0.0.0.0:8700"
)
document_state = """
Invoice Statement.
Base fee: $150.00. Early bird savings applied: $20.00.
Total balance due: $130.00.
Candidates found by parser: ["150.00", "20.00", "130.00"]
"""                       # CLM_BASE_URL (default http://127.0.0.1:8700), CLM_API_KEY
r = client.system_one(
    state=document_state,
    questions={
        "final_total": Choice(
            instructions="Which of the candidate values represents total balance due amount the customer has to pay?",
            criteria={150:"The total balance to be paid is $150.00", 20:"The total balance to be paid is $20.00",130: "The total balance to be paid is $130.00", "None of these":"none"}
        ),
        "has_discount": Noul(
            instructions="Does this invoice specify that a discount was applied?"
        )
    }
)
end=time.time()
print(end-start)             # 0.41022     probability the statement is true
print(r.answers)     # 38 58.1     (106 tokens on a cold cache: option texts are embedded once)
```
